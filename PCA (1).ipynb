```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Load the dataset
iris = load_iris()

# Create a DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Display first 5 rows
print(df.head())
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"]
)

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Original Iris Data")

plt.show()
```

```python
# Create the scaler
scaler = StandardScaler()

# Standardize all 4 features
X_scaled = scaler.fit_transform(df)

print(X_scaled[:5].round(2))
```

```python
# Create PCA with 2 components
pca = PCA(n_components=2)

# Transform 4 features into 2 components
X_pca = pca.fit_transform(X_scaled)

# Display the transformed data
print(X_pca[:5].round(2))
```

```python
print("Original data shape:", X_scaled.shape)
print("PCA data shape:", X_pca.shape)
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1]
)

plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")
plt.title("Iris Data After PCA")

plt.show()
```

```python
# Variance explained by each component
print("Variance explained by PC1 and PC2:")
print(pca.explained_variance_ratio_.round(3))

# Total variance explained
total = sum(pca.explained_variance_ratio_)

print("Total variance retained:",
      round(total * 100, 2), "%")
```
